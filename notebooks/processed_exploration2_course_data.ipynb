Explore courses with null strokes gained data for tournaments

```python
import pandas as pd

course_diff = pd.read_csv("../data/processed/course_difficulty.csv")

# Get the list of null-SG courses
null_courses = course_diff[course_diff['avg_sg_total'].isnull()]['course'].tolist()
print(len(null_courses), "courses with null avg_sg_total")
print(null_courses)
```

```python
raw = pd.read_csv("../data/raw/pga_tour_raw.csv")
raw_post2017 = raw[raw['season'] > 2016]

null_course_rows = raw_post2017[raw_post2017['course'].isin(null_courses)]

# How many rows, and how many are missing sg_total specifically?
print(null_course_rows.shape)
print(null_course_rows['sg_total'].isnull().sum(), "of", len(null_course_rows), "rows missing sg_total")

# Look at a sample of tournament names — alternate formats often have telltale names
null_course_rows[['course', 'tournament name', 'season', 'sg_total']].drop_duplicates(subset=['course', 'tournament name', 'season'])
```
